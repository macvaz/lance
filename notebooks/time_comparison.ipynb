```python
import duckdb
import lance

from constants import DATASET_NAME, INDEX_COLUMN, get_lance_dataset_path

lance_path = get_lance_dataset_path(DATASET_NAME)
ds = lance.dataset(lance_path)

duckdb.sql(f"""
    EXPLAIN ANALYZE
    SELECT *
    FROM ds
    WHERE {INDEX_COLUMN} = -2461439046089301801
""").show()
```

```python
print(ds.scanner(filter=f"`{INDEX_COLUMN}` = -2461439046089301801").explain_plan(verbose=True))
```

```python
scanner = ds.scanner(filter=f"`{INDEX_COLUMN}` = -2461439046089301801")
print(scanner.analyze_plan())
```

```python
scanner = ds.scanner(
    columns=[INDEX_COLUMN, "EventDate", "ClientIP", "RegionID"],
    filter=f"`{INDEX_COLUMN}` = -2461439046089301801",
)
print(scanner.analyze_plan())
```
