```python
import polars as pl

ids = [9110818468285196899, 6308646140879811077]

df = (
    pl.scan_parquet("datasets/hits.parquet")
    .filter(pl.col("WatchID").is_in(ids))
    .collect()
)
print(df)
```

```python
import duckdb, lance

ds = lance.dataset("datasets/hits.lance")

duckdb.sql("""
    EXPLAIN ANALYZE
    SELECT *
    FROM ds
    WHERE WatchID = 9110818468285196899
""").show()
```

```python
print(ds.scanner(filter="`WatchID` = 9110818468285196899").explain_plan(verbose=True))
```

```python
scanner = ds.scanner(filter="`WatchID` = 9110818468285196899")
print(scanner.analyze_plan())
```

```python
scanner = ds.scanner(
    columns=["WatchID", "EventTime", "UserID", "URL"],
    filter="`WatchID` = 9110818468285196899",
)
print(scanner.analyze_plan())
```
